# Writing style fixture

## Purpose and scope

As it stands, this project aims to clarify theoretical assumptions,
methodological limitations, and interpretive boundaries in Raman
spectroscopy and related practices, particularly where theory, data
processing, and modern ML pipelines interact (in which the final one is
more of an after thought, but still good). The goal is not to replace
existing theories or invalidate prior work, though there will be the
case where we have to do it (reduce that to be honest), but to identify
explicit assumptions, structural constraints, paradox between
cross-interpretation, and downstream effectiveness of applications based
on the theoretical core arguments. That said, while deliberately
speaking we can use authoritative and flair, it is often seriously
recommend, of `Q1` journals to not use *certain words* and *logical
structures*, even though it is sound, because it relies on rhetoric,
emotional presentation, or simply *attacks of people* rather than
people’s works. That said, a collection of journals accepting of `Q1`
all of those critiques are available, and in the event of us submitting
to there, it should be not too worry. But simply speak, the margin of
error should be done as recommended below, so to widen the gap.

### Writing principles

The writing principle is as simple as to guide toward neutral tones and
professional analysis, but not hindering thought and logical patterns,
but only *reformulation, presentation* and so on.

#### General

Each theoretical framework, model, or processing step should be
introduced together with: its **domain of validity**, its implicit
*assumptions*, and known *limitations*. When present them and stating
assumptions or theoretical framework, please use **descriptive** as
possible, conditional phrasing (e.g. “valid under”, “commonly assumed”,
“effective representation”) rather than **evaluative** or *normative*
language. This is because words are weapons toward which in physics, the
sensitive people procure resistant to their own works being criticised.
I.e. so is to don’t ‘insult’ them academically. You can refer that to a
list of **neutral words and adjectives**, and styling so on aside from
the below section, using [AIP Style
Manual](https://publishing.aip.org/wp-content/uploads/2021/03/AIP_Style_4thed.pdf).
There are also specific styling for example, instead of
$\alpha$-particle, use $\alpha$ particle without hyphen instead.

Specifically, for an example, Adjectives should be used sparingly and
only when justified. Nature Physics specifically advises:

-   Avoid “very” - it adds syllables but not information
-   Avoid vague superlatives like “ultra” (e.g., “ultrashort”) - be
    specific about actual scales instead
-   Don’t overuse buzzwords like “quantum,” “nano,” or “bio” unless
    they’re central to your research
-   Avoid empty phrases like “Holy Grail” or “opening new avenues”
    unless historically so, but otherwise don’t.

That said, **stronger journals** might want *direct criticism* and so
on, and over-soften anything might prove disastrous. In that case, it is
then recommended for you and later development to choose between make
strong claims about contradictions (just frame them as “theoretical
inconsistencies” not “errors”) or not, Point out practical failures
(link them to reproducibility, not incompetence) or not, and Propose
reforms (as “refinements” or “extensions” rather than replacements) or
not, depends on the severity of the problem and parties involved.

#### No personal attribution of fault

Structural issues are framed as field-level practices, not individual
errors or motivations. Historical sections as specification, not
judgement, thus when writing historical developments, the contents are
and should be described in terms of:

-   What was *assumed* of the old theory, and their assumptions, stated
    explicitly of **neutral and descriptive way**,
-   What was *explained*, using the author’s or the theory’s own voice
    (or explaination),
-   and what was left outside *scope* at the time. Hence, inherent
    at-that-time limitation and falsifiable concepts.

All of which is to reduce attribution to certain people’s lifeline and
so on, since our topic is rather dangerous to tread. However, in certain
cases, you can, because of our venue of submission to be very receptive
to *active criticism* and not flair of softend notice. \#### On critique
and limitations

Critiques should be framed as constraints on use, not declarations of
incorrectness. When possible, link theoretical issues to practical
consequences (e.g. interpretability, reproducibility, ML leakage), not
to philosophical disputes alone. We explicitly acknowledge that many
issues discussed are not resolved within this paper, as it is
*identifying problems* and proposing a rather hasty fashion of a
reformation, or where to focus, where to specify, not *resolving
themselves*. This is important enough, very much so.

When interfacing with ML or data-driven methods, clearly distinguish
between physical signal, preprocessing artifacts (baseline correction
reveals granular of error, or for example, the deviation on harmonic
analysis, etc), and model-induced structure. Avoid language that implies
physical meaning where identifiability is unclear, so is to reduce
claims without metric, and to remove resistance from addressing them in
the wrong manner.

#### Collaborative expectations

Disagreements are expected and documented explicitly, so is to be
resolved; even disagreement of technicality between different team
members is good but must be available for resolution closure. If a claim
feels “too strong”, the first response should be to add conditions, not
to remove the claim. If a section risks overreach, we prefer narrowing
scope over dilution. The goal overall is to make less of unwanted
friction between us and the potential reviewer, but still keeps the
legitimacy of the content. That is why we should also *attribute as much
as citation*, but *key citation and key placement of them in the
document* together for maximum effect.
