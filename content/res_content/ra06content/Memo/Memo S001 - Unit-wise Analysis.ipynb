# Memo S001. Unit-wise Analysis

Just an idea. There maybe several unit-wised bias term we need to
define. For example, it would be: - The procedural bias - bias of
successive operations and counts over procedures. - Unit bias - The bias
of singular unit - this comes in handy for procedural bias. - Path
bias - Normal bias of the connections and topology of the network. -
Error correction bias - The bias of the decision machine (for example,
SGD bias).
