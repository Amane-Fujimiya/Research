# Zotero integration

For streamlining different parts of our research, one of the main
technology - or rather an antique one, but in general standard, is
**citation management**. Toward such end, of between various countries
together in a single research, *Zotero* stood out surprisingly well as
one of the main tool of such. So, this is a quick guide.

First, look into [The Zotero
Documentation](https://www.zotero.org/support/) that is available on
their own website, including both installation and usual minimal working
example. Specifically, after reading installation, refers to [Group
configuration](https://www.zotero.org/support/groups) for references and
information related to our own project and how to retrieve and sync the
library with other team members. Respective online library entry is
available for each research topic, as well as the group itself, which is
available in the respective organization list. For a very slim and kind
of outdated tutorial, see [this one
here](https://libguides.unm.edu/Zotero).
