Let $\mathcal{G}\subseteq[a,b]^{\mathcal{Z}}$ be a set of functions
$\mathcal{Z}\to[a,b]$ where $[a,b]\in \mathbb{R}$ and $a<b$. Let
$$Z_{1},\dots,Z_{n}$$ be i.i.d. random variable on $\mathcal{Z}$
following some distribution $P$. Denote the sample
$$S=(Z_{1},\dots,Z_{n})$$ The *empirical Rademacher complexity* of
$\mathcal{G}$ with respect to the sample $S$ is
$$\hat{\mathfrak{R}}_{S}(\mathcal{G}):= \mathbb{E}_{\sigma} \left[ \underset{g\in \mathcal{G}}{\mathrm{sup}} \frac{1}{n} \sum^{n}_{i=1} \sigma_{i} g(Z_{i}) \right]$$
Where $$\sigma= (\sigma_{1},\dots,\sigma_{n})^{\top}$$ with
$\sigma_{i} \overset{i.i.d.}{\sim} \mathrm{unif}\{ -1,1 \}$. Here
$\sigma_{1},\dots,\sigma_{n}$ are known as the **Rademacher** random
variable. The complexity $\hat{\mathfrak{R}}_{S}(\mathcal{G})$ is random
because of the randomness of $S$. The **Rademacher complexity** is then
defined for $\mathcal{G}$ as:
$$\mathfrak{R}_{n}(\mathcal{G}) = \mathbb{E}_{S}[\hat{\mathfrak{R}}_{S}(\mathcal{G})]$$
Rademacher complexity is sometimes called “Rademacher average”
