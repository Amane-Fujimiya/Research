# Definition

We state the definition of the simple graph: \> \[!def\] Simple Graph
(1.1.1) \> A **simple graph** is a pair $(V,E)$, where $V$ is a finite
set, and where $E$ is a subset of $\mathcal{P}_{2}(V)$

To remind, $\mathcal{P}_{2}(V)$ is the set of all 2-element subsets of
$V$. This is analogous to the edges or connections. Thus, a simple graph
is a pair $(V,E)$ where $V$ is a finite set, and $E$ is a set consisting
of 2-element subsets of $V$. \> \[!example\] Example 1.1.2 \> For any
$n\in \mathbb{N}$, we can define a simple graph $\mathrm{Cop}_{n}$ to be
the pair $(V,E)$ where $V=\{ 1,2\dots,n \}$ and
$$E=\{ \{ u,v \} \in \mathcal{P}_{2}(V) \mid gcd(u,v)= 1\}$$ \> We call
this the **$n$-th coprimality graph**.

The purpose of simple graphs is to encode relations on a finite set –
specifically the kind of relations that are binary (i.e., relate pairs
of elements), symmetric (i.e., mutual) and irreflexive (i.e., an element
cannot be related to itself).
