Flattening is the simple operation of compressing multi-dimensional
matrix (like adjacent and embedding matrix) into lower-dimensional
version. The simple example is the MLP-Graph conversion. To generate an
embedding of an entire graph we could simply flatten the adjacency
matrix and feed the result to a multi-layer perceptron (MLP):
$$z_{\mathcal{G}}=\mathrm{MLP}(\mathbf{A}[1]\oplus \mathbf{A}[2]\oplus\dots \oplus \mathbf{A}[\lvert \mathcal{V} \rvert ])$$
Where $\mathbf{A}[i]\in\mathbb{R}^{\lvert \mathcal{V} \rvert}$ denotes a
row of the adjacency matrix, and $\oplus$ is the matrix-vector
concatenation operation.
