# Definitions

A **simple graph** $G$ consists of a non-empty finite set $V(G)$ of
elements called **vertices** or **nodes**, and a finite set $E(G)$ of
distinct unordered pairs of distinct elements of $V(G)$ called
**edges**. We called $V(G)$ the **vertex set** and $E(G)$ the **edge
set** of $G$.

An edge $\{ v,w \}$ is said to **join** the vertices $v$ and $w$ and is
usually abbreviated to $vw$.

A **general graph** is the generalization of simple graph, with
multi-edge connections and **loops** for any vertices. Thus a **graph**
$G$ consists of a non-empty finite set $V(G)$ of elements called
**vertices** and a finite family $E(G)$ of unordered pairs of elements
of $V(G)$ called **edges**.

The language of graph theory is not standard - all authors have their
own terminology. Some use the term ‘graph’ for what we call a simple
graph, or for a graph with directed edges, or for a graph with
infinitely many vertices or edges. \_\_\_
