W say that two vertices $v$ and $w$ of a graph $G$ are **adjacent** if
there is an edge $vw$ joining them, and the vertices $v$ and $w$ are
then **incident** with such an edge. Similarly, two distinct edges $e$
and $f$ are **adjacent** if they have a vertex in common.
\![\[Research/Research Repository III/GNN and Statistical Learning
(2024-2025)/Theory Documentation/Pasted image
20241012230214.png\|400\]\] The *degree* of a vertex $v\in G$ is the
number of edges incident with $v$, and is written $deg(v)$. In
calculating the degree of $v$, we usually make the convention that a
loop at $v$ contributes 2 (rather than 1) to the degree of $v$. A vertex
of degree 0 is an **isolated vertices** and a vertex of degree 1 is an
**end-vertex**.

The **degree sequence** of a graph consists of the degrees written in
increasing order, with repeats where necessary.
