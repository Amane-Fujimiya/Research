Double descent lies in the realm of what is considered **error
analysis**, performance evaluation, *inference testing*, or in short,
within the end-to-end pipeline of a specific model, the act of examining
the return behaviours of the model at inference time, normally by
empirical means. By that, in literature, double descent and its related
empirical phenomena is treated via *statistical learning theory* (Vapnik
-Chervonenkis), **computational learning theory**, and, by model design,
the *bias-variance tradeoff* (Geman et al.), a concept taken from
statistics to machine learning model. Setting up the problem requires
all of them. \# I. STL and COLT Statistical learning theory analyze the
machine learning problem using statistical analysis of the process,
providing a framework for studying the problem of inference,
predictions, and decisions or constructions of model from a set of data.
The goal, in such, would be to formalize notions that we taken for
granted - *generalization*, *bias-variance*, *underfitting &
overfitting*, and else.

We formulate the problem of machine learning in this framework by the
hypothesis-concept model, or the function estimation model. In parts, we
are considering the process of behavioural matching, with the goal to
then not just match the observed behaviours, but also figure out the
underlying principle and behavioural control details. Thereby, a
**statistical learning problem** often consists of: 1. The **observed
space** $\mathcal{D}[\mathbb{R}^{n}]$ with observations, or *data
points* of desired concept. 2. The **hypothesis** or model function
$h[\Theta\in \Lambda](\mathbf{x})\in \mathcal{H}$, where $\Theta$ is the
set of concurrent descriptions of the hypothesis’s function, and
$\Lambda$ is the set of all such set of descriptions; $\mathbf{x}$ is
the **background of the observable**. 3. The **concept** or target
function $c[\alpha](\mathbf{x}')$ with its own concurrent descriptions,
belong to $\Lambda'$, and $\mathbf{x}'$ is its own background. The
**problem** of learning is then of choosing from the given set of
hypothesis $\mathcal{H}$ the hypothesis that first *imitate* the
concept, and second is to predict the concept in the best possible way.
Taking computational complexities, existence of its solution, and
computable algorithm within polynomial time, for example, then we have
**computational learning theory** (COLT) - in which case is simply the
extension of statistical learning theory.
