Any function that takes an adjacency matrix $\mathbf{A}$ as input should
ideally satisfy the following properties, known as **permutation
invariance and equivariance**:
$$f(\mathbf{PAP}^{\top})=f(\mathbf{A}),\quad f(\mathbf{PAP}^{\top})=\mathbf{P}f(\mathbf{A})$$
where $\mathbf{P}$ is a permutation matrix. *Permutation invariance*
means that the function does not depend on the arbitrary ordering of the
rows/columns in the adjacency matrix, while permutation equivariance
means that the output of $f$ is permuted in a consistent way when we
permute the adjacency matrix. Ensuring invariance or equivariance is a
key challenge when we are learning over graphs, and any system of GNN
should ideally satisfies these fundamental properties.
