We notice that a graph $G$ has a **maximum degree** (the largest degree
of any vertex) and a **minimum degree** (the smallest degree of any
vertex). Denoted $\Delta(G)$ and $\delta(G)$ accordingly, we then turn
into the first lemma that counts the total amount of graph connections,
the *Handshaking lemma* (Leonhard Euler, 1736). \> \[!lem\] The
Handshaking Lemma \> In any graph $G$, the sum of all the vertex-degree
is an even number. Or, the vertex degree add up to *twice* the number of
edges: $$\sum_{v\in V(G)}\mathrm{deg}_{G}(v)= 2\lvert E(G) \rvert $$ \#
Proof We present the proof of induction. The base case here is $m=0$.
The graph with no edge then $\mathrm{deg}(v)=0$.

Assume that the degree sum formula holds for all $(m-1)$-edge graphs.
Let $G$ be a graph with $m\geq 1$ edges, and let $xy$ be any edge of
$G$. We can apply the inductive hypothesis to $G-xy$ (the graph we get
from **deleting** $xy$ from $G$), a graph with $m-1$ edges.

Both $x$ and $y$ have one extra incident edge in $G$ they don’t have in
$G-xy$ - the edge $xy$ itself. Hence
$$\mathrm{deg}_{G}(x)=1+\mathrm{deg}_{G-xy}(x)$$ and
$$\mathrm{deg}_{G}(y)=1+\mathrm{deg}_{G-xy}(y)$$ For any other vertex
$v$, $G-xy$ and $G$ have the same number of edges, so we have
$$\mathrm{deg}_{G}(v)=\mathrm{deg}_{G-xy}(v)$$ Furthermore, $G-xy$ and
$G$ have the same set of vertices. So if we add up the vertex degrees in
$G-xy$ and $G$, the result is that
$$\sum_{v\in V(G)} \mathrm{deg}_{G}(v)= 2+ \sum_{V\in V(G-xy)}\mathrm{deg}_{G-xy}(v)$$
Applying the inductive hypothesis, we get that the degree sum in $G-xy$
is $2(m-1)$, so the degree sum in $G$ is $2(m-1)+2=2m$. By induction,
the degree sum formula holds all graphs.
