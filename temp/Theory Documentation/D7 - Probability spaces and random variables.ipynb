This provides us the definitions on **probability spaces** and **random
variables**

> \[!def\] Sample spaces For a random experiment, the sample space
> $\Omega$ is the set of possible outcomes of the random experiment.

> \[!def\] Events An event of a random variable with sample space
> $\Omega$ is a subset $E\subset \Omega$.

------------------------------------------------------------------------

> \[!def\] Probability spaces and random variables Let
> $(\Omega,\mathcal{H},\mathbb{P})$ be a probability space. The set
> $\Omega$ is called the **sample space**; its elements are called
> *outcomes*. The $\sigma$-algebra $\mathcal{H}$ may be called the grand
> history; its elements are called events. We repeat the properties of
> the probability measure $\mathbb{P}$; all sets here are events: 1.
> Norming: $\mathbb{P}(\varnothing)=0,\quad\mathbb{P}(\Omega)=1$ 2.
> Monotonicity: $H\subset K\implies \mathbb{P}(H)\leq \mathbb{P}(K)$ 3.
> Finite additivity:
> $H\cap K=\varnothing\implies \mathbb{P}(H\cup K)=\mathbb{P}(H)+\mathbb{P}(K)$
> 4. Countable additivity: $(H_{n})$ disjointed implies
> $$\mathbb{P}\left( \bigcup_{n} H_{n} \right)=\sum_{i\leq n}\mathbb{P}(H_{n})$$
> 5. Sequential Continuity:
> $H_{n}\nearrow H\implies \mathbb{P}(H_{n})\nearrow\mathbb{P}(H)$ and
> $H_{n}\searrow H\implies \mathbb{P}(H_{n})\searrow\mathbb{P}(H)$ 6.
> Boole’s inequality:
> $$\mathbb{P}\left( \bigcup_{n} H_{n} \right)\leq \sum_{n}\mathbb{P}(H_{n})$$
