# Isomorphism

Two graphs $G_{1}$ and $G_{2}$ are isomorphic if there is a one-one
correspondence between the vertices of $G_{1}$ and those of $G_{2}$ such
that the number of edges joining any two vertices of $G_{1}$ is equal to
the number of edges joining the corresponding vertices of $G_{2}$.

For many problems, the labels on the vertices are unnecessary and we
drop them. We then say that two ‘unlabelled graphs’ are isomorphic if we
can assign labels so that the resulting ‘labelled graphs’ are
isomorphic,. \# Connectedness We can combine two graphs to make a larger
graph. If the two graphs are $G_{1}=(V(G_{1}),E(G_{1}))$ and
$G_{2}=(V(G_{2}),E(G_{2}))$ where $V(G_{1})$ and $V(G_{2})$ are
disjoint, then their **union** $G_{1}\cup G_{2}$ is the graph with
vertex set $V(G_{1})\cup V(G_{2})$ and edge family
$E(G_{1})\cup E(G_{2})$.

A graph is **connected** if it cannot be expressed as union of two
graphs, and disconnected otherwise.
