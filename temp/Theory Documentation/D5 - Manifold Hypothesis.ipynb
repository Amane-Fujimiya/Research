# 

The manifold hypothesis is a widely accepted tenet which posits that: \>
\[!quote\] Manifold Hypothesis - Lawrence Cayton \> \[…\] the
dimensionality of many data sets is only artificially high; though each
data point consists of perhaps thousands of features, it may be
described as a function of only a few underlying parameters. That is,
the data points are actually samples from a low-dimensional manifold
that is embedded in a high-dimensional space.

Essentially, high-dimensional data sets that occur in the real world
actually lie along low-dimensional latent manifold inside that
high-dimensional spaces. As a consequence of the manifold hypothesis,
many data sets that appear to initially require many variables to
describe, can actually be described by a comparatively small number of
variables.
